```python
import pandas as pd
import numpy as np
#creating series from array
arr=np.array(['P','a','n','d','a','s'])
a=pd.Series(arr)
print(a)
```

```python
#Series from dictionary
arr={'X':0.,'Y':1.,'Z':2.}
b=pd.Series(arr)
print("\n\n Series fromm dictionary:\n")
print(b)
```

```python
#Series using scalar
X=pd.Series(4,index=[0,1,2,3])
print("\n Series using Scalar\n")
print(X)
```

```python
#Series through index
X=pd.Series([1,2,3],index=['a','b','c'])
print("\n Series through index")
print(X)
```

```python
a=pd.Series(data=[1,2,3,4])
print("\n a Series:\n",a)
print("\n Index:\n",a.index)
print("\n Values:\n",a.values)
print("\n Shape:",a.shape)
print("\n Dimension:",a.ndim)
print("\n Size:",a.size)
```
